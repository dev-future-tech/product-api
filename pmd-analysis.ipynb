```python
import pandas as pd

PMD_CSV_FILE=r'report.csv'
```

```python
coverage = pd.read_csv(PMD_CSV_FILE)
prioritized = coverage.sort_values(by=["Priority"], ascending=True)

prioritized.head()

```

```python
pd.set_option('display.max_colwidth', None)
prioritized_categories = coverage.groupby(["Description", "Priority"])["Description"].count().reset_index(name="count")

prioritized_categories.sort_values(by=["count"], ascending=False).head()
# coverage.groupby(["Description", "Priority"])["Description"].count().reset_index(name="count").sort_values(by=["count"], ascending=False)
```
